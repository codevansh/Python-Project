```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("housing.csv")
```

```python
df
```

```python
df["income_cat"] = pd.cut(df['median_income'], bins =[0,1.5,3.0,4.5,6.0, np.inf], labels=[1,2,3,4,5])
```

```python
from sklearn.model_selection import StratifiedShuffleSplit
split = StratifiedShuffleSplit(n_splits=1,test_size=0.2,random_state=42)
for train_index,test_index in split.split(df,df['income_cat']):
  strata_train_set = df.loc[train_index]
  strata_test_set = df.loc[test_index]
```

```python
df = strata_train_set.copy()
```

```python
df
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
```

```python
my_Pipeline = Pipeline([
  ("impute", SimpleImputer(strategy="median")),
   ('standardize',StandardScaler())
])

# pipelines are used to implement too many updations on the data at once in a single cell.
```

```python
housing = df.drop('ocean_proximity',axis=1)
```

```python
housing
```

```python
my_Pipeline.fit_transform(housing)
```
